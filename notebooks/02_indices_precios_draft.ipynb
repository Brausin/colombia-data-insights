El **Índice de Precios al Productor (IPP)** mide la variación de precios de bienes en la etapa de producción, antes de llegar al consumidor final. A menudo **el IPP anticipa movimientos del IPC** porque los aumentos de costos de producción tarde o temprano se trasladan a los precios minoristas.
